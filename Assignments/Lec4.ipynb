## *Recreate the Following Effects*



```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('../Assets/opencv_logo.png')

dst = cv2.boxFilter(img, -1, (30,30))
plt.subplot(121),plt.imshow(img),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(dst),plt.title('Averaging(Box) Filter')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
img = cv2.imread('../Assets/opencv_logo.png')

dst = cv2.GaussianBlur(img, (77, 77), 0)
plt.subplot(121),plt.imshow(img),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(dst),plt.title('Gaussian Blur Filter')
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
img = cv2.imread('../Assets/Assignment2-Noise.png')

dst = cv2.medianBlur(img, 11)
plt.subplot(121),plt.imshow(img),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(dst),plt.title('Median Filter')
plt.xticks([]), plt.yticks([])
plt.show()
```
