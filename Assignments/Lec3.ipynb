## Transparency Experiment

Use **4 subplots** to show different transparency levels of your ID image.

---


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
rectangleColor = (0, 255, 255)
background = np.zeros((480, 640, 3), dtype=np.uint8)

image = cv2.rectangle(background,(100,100),(150,200),rectangleColor,1)
image = cv2.rectangle(image,(200,200),(300,300),rectangleColor,1)
image = cv2.rectangle(image,(250,320),(400,400),rectangleColor,1)

cv2.imshow('image', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
myID = cv2.imread('../Assets/ID.png')

myID = cv2.circle(myID, (100, 100), 25, (0, 0, 255), 2)
myID = cv2.circle(myID, (100, 280), 25, (0, 0, 255), 2)
myID = cv2.circle(myID, (350, 100), 25, (0, 0, 255), 2)
myID = cv2.circle(myID, (350, 280), 25, (0, 0, 255), 2)

cv2.putText(myID, 'Mahmoud', (150, 80), cv2.FONT_HERSHEY_COMPLEX, 1, (0, 0, 0), 2, cv2.LINE_AA)
myID = cv2.line(myID, (150, 100), (300, 100), (0, 0, 0), 2)

cv2.imshow('image', myID)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

alphas = [0.25, 0.5, 0.75, 1.0]  # Transparency levels
for ax, alpha in zip(axes, alphas):
    ax.imshow(cv2.cvtColor(myID, cv2.COLOR_BGR2RGB), alpha=alpha)
    ax.set_title(f'Transparency: {alpha}')
    ax.axis('off')

plt.tight_layout()
plt.show()
```
